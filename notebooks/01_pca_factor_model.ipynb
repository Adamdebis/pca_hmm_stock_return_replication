# 01 - PCA factor model

PCA rotates the standardized return matrix into orthogonal directions ordered by explained variance. We retain the smallest number of directions explaining at least 1-p of the variance.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
from src.data import make_synthetic_returns
from src.pca_hmm import fit_pca
returns = make_synthetic_returns(periods=320, assets=24, seed=42)
pca = fit_pca(returns.to_numpy(), noise_fraction=0.15)
print('components:', pca.n_components)
print('variance retained:', pca.explained_variance_ratio[:pca.n_components].sum())
plt.plot(pca.cumulative_explained_variance, marker='o')
plt.axhline(0.85, color='red', linestyle='--')
plt.xlabel('Component')
plt.ylabel('Cumulative explained variance')
plt.show()
```
