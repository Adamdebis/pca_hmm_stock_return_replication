# 02 - Gaussian hidden Markov model

The HMM observes PCA factor returns but not the market-regime label. Baum-Welch estimates the transition matrix and Gaussian state parameters; Viterbi identifies the most likely current state.

```python
from pathlib import Path
import sys
import numpy as np
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
from src.data import make_synthetic_returns
from src.pca_hmm import fit_pca, select_hmm
returns = make_synthetic_returns(periods=320, assets=24, seed=42)
pca = fit_pca(returns.to_numpy(), noise_fraction=0.15)
hmm, aic_table = select_hmm(pca.factors, state_candidates=(2, 3, 4), max_iter=35)
print(aic_table)
print('selected states:', hmm.n_states)
print('transition rows:', hmm.transition_matrix_.sum(axis=1))
print('next factor forecast:', hmm.forecast_next(pca.factors))
```
